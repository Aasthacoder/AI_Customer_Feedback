##### 1. Problem Prioritization

The AI-assisted analysis identified five candidate customer problems:

- Purchase disputes / refunds
- Late payments / billing errors
- Promotions / rewards
- Fees / interest
- Fraud / unauthorized charges

These candidate problems are evaluated using three factors:

- Evidence from the customer-feedback analysis
- Potential customer impact
- Potential for product intervention

##### 2. Comparing Candidate Problems

| Candidate problem | Evidence from analysis | Potential customer impact | Product opportunity |
|---|---|---|---|
| Purchase disputes / refunds | Strong | High | High |
| Late payments / billing errors | Moderate | High | Medium |
| Promotions / rewards | Moderate | Medium | High |
| Fees / interest | Strong | High | Medium |
| Fraud / unauthorized charges | Moderate | Very high | High |

This comparison is directional rather than a precise numerical score. The available complaint data provides evidence of recurring problems but does not directly measure customer impact or market demand.

##### 3. Selected Customer Problem

##### Selected Problem

**Customers experience difficulty getting purchase disputes resolved.**

Purchase-dispute complaints formed one of the largest identified themes in the analysis. The relevant cluster also showed strong concentration around purchase-related complaints.

The problem is therefore selected as the starting point for further product exploration.

##### 4. Evidence and Problem Hypothesis

The complaint narratives provide examples of customers reporting unresolved disputes, difficulties obtaining resolution, and continued interaction with companies regarding purchase-related problems.

However, the dataset does not directly measure the customer's complete dispute journey or the number of contacts made for each dispute.

##### Problem Hypothesis

Customers may experience uncertainty and friction while trying to resolve a purchase dispute, particularly when they do not understand the status, outcome, or next steps.

##### 5. Product Opportunity

Based on the problem hypothesis, there is an opportunity to make the purchase-dispute process easier for customers to understand and follow.

The initial product opportunity is to improve visibility into:

- dispute status,
- information required from the customer,
- next steps,
- and dispute outcomes.

This is an opportunity hypothesis and would require customer testing to confirm that these improvements address the underlying frustration.

##### 6. Root Cause Hypothesis

The complaint data suggests that some customers experience friction after raising a purchase dispute.

The available narratives point toward issues such as unresolved disputes, difficulty obtaining a resolution, and continued interaction with the company.

However, the dataset does not contain the complete customer journey, so the underlying root cause cannot be confirmed from complaint data alone.

##### Root Cause Hypothesis

Customers may lack sufficient clarity about the progress and resolution of their purchase disputes, which can contribute to uncertainty and repeated effort.

##### 7. Solution Direction

The solution should focus on making the existing purchase-dispute process easier to understand and follow.

The initial solution direction is to provide customers with clearer information about:

- Current dispute status
- Information required from the customer
- Next steps
- Dispute outcome

##### 8. MVP Definition

MVP — Dispute Tracker

The MVP will provide a single place where customers can:

- View the current dispute status
- See whether additional information is required
- Understand the next step
- View the dispute outcome

The MVP focuses on improving transparency rather than replacing the existing dispute investigation process.

##### 9. MVP Scope

Included

- Dispute status tracking
- Required-information notifications
- Next-step guidance
- Clear dispute outcome information

Not Included

- Automatic approval or rejection of disputes
- Automated fraud detection
- Replacement of human investigation
- A new dispute-submission system

Keeping these capabilities outside the MVP allows the first version to focus on the identified transparency problem.

##### 10. Success Metrics

The primary goal of the MVP is to reduce unnecessary effort for customers while they are waiting for a purchase dispute to be resolved.

**Primary Metric** - **Repeat contact rate for the same dispute**

**Supporting Metrics**

- Average dispute resolution time
- Percentage of disputes requiring additional customer contact
- Customer satisfaction after dispute resolution

**Guardrail Metric** - **Dispute outcome quality**

A reduction in customer contact should not come at the cost of poorer dispute outcomes or reduced access to support.

##### 11. Experiment / Validation Plan

##### Hypothesis

If customers receive clearer information about their purchase dispute status, required information, and next steps, then unnecessary repeated contact about the same dispute will decrease.

##### Experiment

- **Control:** Existing purchase-dispute experience
- **Treatment:** Existing experience with the Dispute Tracker
- **Primary metric:** Repeat contact rate for the same dispute
- **Supporting metrics:** Resolution time and customer satisfaction
- **Guardrail:** Dispute outcome quality

The experiment should compare the existing experience with the MVP before considering a wider rollout.

##### 12. Rollout Plan

The MVP should be introduced gradually.

| Stage | Approach | Goal |
|---|---|---|
| 1. Usability testing | Test the experience with a small group of relevant customers. | Identify usability problems. |
| 2. Limited rollout | Release the MVP to a small percentage of eligible customers. | Measure the defined success metrics. |
| 3. Evaluate results | Compare the MVP experience with the existing experience. | Determine whether the MVP improves the customer experience. |
| 4. Wider rollout | Expand availability if the results meet the success criteria. | Scale the validated experience. |

##### 13. Product Roadmap

| Phase | Focus | Key outcome |
|---|---|---|
| Phase 1 — MVP | Status, required information, and next-step guidance | Customers can follow an active dispute more easily. |
| Phase 2 — Communication | Clearer updates and outcome explanations | Customers receive better information throughout the dispute process. |
| Phase 3 — AI assistance | Simpler explanations of complex dispute information | Customers can understand dispute information more easily. |

The roadmap prioritizes validation of the core experience before introducing additional AI capabilities.

##### 14. Final Product Hypothesis

If customers can clearly understand the status, required actions, next steps, and outcome of their purchase dispute, then unnecessary repeated contact about the same dispute should decrease.

The MVP should be considered successful if it improves the primary metric without negatively affecting dispute outcomes or customer access to support.

#### 15. Conclusion

The project used real CFPB consumer complaint data to identify recurring customer problems and applied AI-assisted analysis to discover and validate customer themes.

Purchase-dispute resolution was selected as the product opportunity based on the strength of the available evidence and its potential customer impact.

The proposed MVP, Dispute Tracker, focuses on improving transparency during the existing dispute process.

The next step would be to test the MVP with real customers and determine whether it reduces unnecessary repeated contact while maintaining appropriate dispute outcomes.
